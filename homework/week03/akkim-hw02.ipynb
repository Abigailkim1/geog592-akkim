## Reading the original CSV file

Variable: Average production yield

```python
import csv 

filename = "tabela1613.csv" 

# Argument "r" stands for read. 
# Had to specify the encoding.
# Total does not have any values. 
# Only need maybe Code and Avg_yield
with open(filename, "r", encoding="utf-8") as file: 
    fieldnames = ['Level', 'Code', 'Municipality', "Total", "Unit0", "Avg_yield", "Unit"]
    reader = csv.DictReader(file, fieldnames)
    # Skipping the first 5 rows because those are just notes. 
    for i in range(0,5):
        next(reader)
    rows = list(reader)

print(f"Records found: {len(rows)}") 
```

## Selecting columns

Selecting just Code and Avg_yield columns

```python
rows_selected = []

# 'row' accesses dictionary for each row. 
for row in rows_filtered: 
    newDict = {'Code': row['Code'],
               'Avg_yield': row['Avg_yield']}
    rows_selected.append(newDict)

head(rows_selected)
```

## Checking the selected data for problems

### Checking for missing values

Nulls seem to be indicated by '-' or '...'

```python
# Function for checking how many rows are removed.
# I was going to use this for checking every time I removed rows, 
# but the rest of the data ended up being fine. 

def rows_removed(list1, list2):
    print(f"This operation removed {len(list1) - len(list2)} rows. There are now {len(list2)} rows.")
```

```python
# This feels very inefficient, 
# but I can't use .pop() without causing an iteration error (since the size of the list changes)

rows1 = []

for row in rows_selected: 
    if row['Code'] != '-' and row['Code'] != '...':
        rows1.append(row)

rows2 = []

for row in rows1:
    if row['Avg_yield'] != '-' and row['Avg_yield'] != '...':
            rows2.append(row)

for i in range(len(rows2)): 
    print(rows2[i])

rows_cleaned = rows2
```

```python
rows_removed(rows1, rows2)
```

Looks good so far. 

### Checking for non-numeric values in avg_yield

```python
import re 

weird_counter = 0
for row in rows_cleaned: 
    value = row['Avg_yield']

    if re.search(r"[^A-Za-z0-9 ]", value):
        weird_counter += 1
        print(f"Suspicious value found at {row['Code']}: {row['Avg_yield']}")

if weird_counter == 0:
    print("No suspicious values found!")
```

### Converting to numeric data type

Check first that everything can be an int, then convert. 

```python
def cannot_be_int(number):
    try:
        int(number)
        return False
    except ValueError: 
        return True
```

```python
error_counter = 0 

for row in rows_cleaned: 
    if cannot_be_int(row['Code']):
        error_counter += 1
        print(f"Code {row['Code']} cannot be converted to an integer.")
    if cannot_be_int(row['Avg_yield']):
        error_counter += 1
        print(f"Avg_yield {row['Avg_yield']} at Code {row['Code']} cannot be converted to an integer.")

if error_counter == 0: 
    print("No problems converting to integers!")
```

```python
for row in rows_cleaned: 
    row['Code'] = int(row['Code'])
    row['Avg_yield'] = int(row['Avg_yield'])

head(rows_cleaned)
```

```python
bad_ints = 0

for row in rows_cleaned: 
    if type(row['Code']) != int:
        bad_ints += 1
        print(f"{row['Code']} is not an integer!")
    if type(row['Avg_yield']) != int:
            bad_ints += 1
            print(f"{row['Avg_yield']} at Code {row['Code']} is not an integer!")

if bad_ints == 0: 
    print("All values are integers!")
```

```python
for row in rows_cleaned:
    print(row)
```

## Export the cleaned data as a new CSV file

```python
# "w" stands for write 

fieldnames = ["Code", "Avg_yield"]
with open("rows_cleaned.csv", "w", newline="") as file: 
    writer = csv.DictWriter(file, fieldnames)
    writer.writeheader()
    writer.writerows(rows_cleaned)
```
