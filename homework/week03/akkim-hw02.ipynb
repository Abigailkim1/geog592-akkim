## Reading the original CSV file

Variable: Average production yield

```python
import csv 

filename = "tabela1613.csv" 

# Argument "r" stands for read. 
# Had to specify the encoding.
# Total does not have any values. 
# Only need maybe Code, Municipality, and Avg_yield
with open(filename, "r", encoding="utf-8") as file: 
    fieldnames = ['Level', 'Code', 'Municipality', "Total", "Unit0", "Avg_yield", "Unit"]
    reader = csv.DictReader(file, fieldnames)
    # Skipping the first 5 rows because those are just notes. 
    for i in range(0,5):
        next(reader)
    rows = list(reader)

print(f"Records found: {len(rows)}") 
```
