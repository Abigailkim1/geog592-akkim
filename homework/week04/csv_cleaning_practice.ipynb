## Importing the csv

```python
import csv 

with open("data/bmf_master_NC.csv", "r") as file:
    reader = csv.DictReader(file)
    rows = list(reader)
```

```python
for i in rows[0]:
    print(i)
```

## Filtering columns 

```python
# New list holding data. 
selected_columns = []
# Columns names to extract 
col_names = ["ein", "org_name_display", "org_addr_full", "ntee_code_major_group", "ntee_code_definition", "last_year_in_bmf", "geo_match_addr", "geo_lat", "geo_lon", "geo_county", "geo_is_geocoded", "org_addr_is_po_box"]

for row in rows:
    new_dict = {}
    for i in col_names: 
        new_dict[i] = row[i]
    selected_columns.append(new_dict)
```

```python
def head(list): 
    for i in range(5): 
        print(list[i])
```

```python
head(selected_columns)
```

## Filtering rows 

Keep only those that are geocoded, not a PO box, last year 2026, and are in Wake County.\
Maybe also check org address against geocoded address.

```python
def rows_removed(list1, list2):
    print(f"This operation removed {len(list1) - len(list2)} rows. There are now {len(list2)} rows.")
```

```python
geocoded = []

for row in selected_columns: 
    if row['geo_is_geocoded'] == "TRUE":
        geocoded.append(row)

rows_removed(selected_columns, geocoded)
head(geocoded)
```

```python
for row in geocoded: 
    if row['org_addr_is_po_box'] == '':
        print(row['ein'])
```

```python
def filter(input, column, value):
    new_list = []
    for row in input: 
        if row[column] == value: 
            new_list.append(row)
    rows_removed(input, new_list)
    head(new_list)
    return new_list
```

```python
not_PO = filter(geocoded, 'org_addr_is_po_box', 'FALSE')
current = filter(not_PO, 'last_year_in_bmf', '2026')
wake = filter(current, 'geo_county', 'Wake County')
```

## Filtering the columns again 

```python
# New list holding data. 
bmf_filtered = []
# Columns names to extract 
col_names = ["ein", "org_name_display", "org_addr_full", "ntee_code_major_group", "ntee_code_definition", "geo_match_addr", "geo_county", "geo_lat", "geo_lon", ]

for row in wake:
    new_dict = {}
    for i in col_names: 
        new_dict[i] = row[i]
    bmf_filtered.append(new_dict)

head(bmf_filtered)
print(len(bmf_filtered))
```

## Converting the coordinates to numerical

```python
def cannot_be_float(number):
    try:
        float(number)
        return False
    except ValueError: 
        return True
```

```python
error_counter = 0

for row in bmf_filtered: 
    if cannot_be_float(row['geo_lat']):
        error_counter += 1
        print(f"{row['ein']} lat cannot be converted.")
    if cannot_be_float(row['geo_lon']):
        error_counter += 1
        print(f"{row['ein']} cannot be converted.")

if error_counter == 0:
    print("No errors!")
```

```python
for row in bmf_filtered: 
    row['geo_lat'] = float(row['geo_lat'])
    row['geo_lon'] = float(row['geo_lon'])

head(bmf_filtered)
```

## Exporting to CSV

```python
fieldnames = []
for col_name in bmf_filtered[0]: 
    fieldnames.append(col_name)

print(fieldnames)
```

```python
with open("bmf_filtered.csv", "w", newline="") as file: 
    writer = csv.DictWriter(file, fieldnames)
    writer.writeheader()
    writer.writerows(bmf_filtered)
```
