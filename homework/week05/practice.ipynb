````python
import numpy as np
print(np.__version__)
````

````python
sites = np.array([
    [681000, 3975000],
    [681400, 3975400],
    [681900, 3975900],
    [682300, 3976200],
    [680700, 3975700]
])

sites
````

````python
bounding = ((sites[:, 0] > 681300) &
            (sites[:, 0] < 682000) &
            (sites[:, 1] > 3975300) &
            (sites[:, 1] < 3976000)
            )

selected = sites[bounding]
selected
````
