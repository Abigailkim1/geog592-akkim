# Pandas summary 

```python
import pandas as pd
print(pd.__version__)
```

```python
# Reading in a csv and inspecting the dataframe. 
# Printing the first 5 rows.
# Must use pd. to access pandas module. 
# After that, the object is from a pandas class, 
# so can just call the method on the object (dataframe in example)
df = pd.read_csv("physical_NC_with_an_error.csv")
df.head()
```

```python
# Printing the last 5 rows. 
df.tail()
```

```python
# Overview of entire dataframe
df.info
```

```python
# Subsetting a specific column. 
# Returns a pandas series.
df['LocationName']
```

```python
# Selecting multiple columns.
# Use a list of column names.
df[['LocationName', 'Year', 'Category']]
```

```python
# Find the type of objects 
print(type(df))
print(type(df['Year']))
print(type(df))
```

```python
# dtype attribute, for seeing data types for whole data frame.
df.dtypes
```

```python
# Attribute that holds the column names of the dataframe
df.columns 
```

```python
# Can use to replace the column names 
parks = pd.DataFrame({"Park":["Oak","Pine","Maple","Cedar"], "Area":[12,25,18,40], "Visitors":[120,250,180,310]})
parks.columns = ["Parks", "Acreage", "People"]
parks.columns
```

```python
# Selecting rows. 
df.iloc[0:4]
```

```python
# index is an attribute of the df object 
# The default is a series of number starting from 0

df.index
```

```python
# Selecting rows and columns with iloc
df[0:4][['LocationName','StateDesc']]
```

```python
# Filtering by conditions in rows 
df[df['TotalPopulation']>'150,000']
```

```python
# Multiple conditions 
df[
    (df['TotalPopulation']>'150,000') &
    (df['TotalPop18plus'] >'150,000')
]
```

```python
# Calling a function from the pandas package. 
# Coerce replaces errors with NaN
df['TotalPopulation'] = pd.to_numeric(
    df['TotalPopulation'].str.replace(",", "", regex=False), 
    errors='coerce')
```

```python
# Check for NaNs
# Returns series of boolean values
df['TotalPopulation'].isna().head()
```

```python
# Count number of NaNs
df['TotalPopulation'].isna().sum()
```

```python
# Drop na drops rows
df.dropna(subset='TotalPopulation')

# Fill na replaces the na values with another value 
df.fillna(0)
```
